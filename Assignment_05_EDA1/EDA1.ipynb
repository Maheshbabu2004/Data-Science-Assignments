```python
# Exploratory Data Analysis (EDA) on Cardiotocographic Dataset
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset
df = pd.read_csv("Cardiotocographic.csv")
```

```python
# Display first 5 rows
print("First 5 Rows:")
print(df.head())
```

```python
# Dataset Information
print("\nDataset Info:")
print(df.info())
```

```python
# Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Handle Missing Values
df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
# Check Duplicate Records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# Statistical Summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Mean and Median
print("\nMean Values:")
print(df.mean(numeric_only=True))

print("\nMedian Values:")
print(df.median(numeric_only=True))
```

```python
# -----------------------------
# Outlier Detection using Boxplots
# -----------------------------
plt.figure(figsize=(15, 8))
df.boxplot(rot=90)
plt.title("Boxplot of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# Histograms
# -----------------------------
df.hist(figsize=(15, 12), bins=20)
plt.suptitle("Histograms of Numerical Features")
plt.show()
```

```python
# -----------------------------
# Correlation Heatmap
# -----------------------------
plt.figure(figsize=(12, 8))
corr_matrix = df.corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# -----------------------------
# Pair Plot
# -----------------------------
sns.pairplot(df[['LB','AC','FM','UC','ASTV','MSTV','NSP']])
plt.show()

# -----------------------------
# Violin Plot for Target Variable
# -----------------------------
plt.figure(figsize=(8, 5))
sns.violinplot(x='NSP', y='LB', data=df)
plt.title("Baseline FHR Distribution by NSP")
plt.show()
```

```python

# -----------------------------
# Scatter Plot
# -----------------------------
plt.figure(figsize=(8, 5))
sns.scatterplot(x='ASTV', y='MSTV', hue='NSP', data=df)
plt.title("ASTV vs MSTV")
plt.show()
```

```python
# -----------------------------
# IQR Method for Outlier Detection
# -----------------------------
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

outliers = ((df < (Q1 - 1.5 * IQR)) |
            (df > (Q3 + 1.5 * IQR))).sum()

print("\nOutliers in Each Column:")
print(outliers)
```
